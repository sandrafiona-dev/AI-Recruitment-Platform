# 02 — Data Cleaning

## Purpose
Define a reproducible approach for reviewing and preparing approved datasets before exploratory analysis.

## Planned Cleaning Activities
- Validate identifiers, required fields, and cross-table relationships.
- Standardize text, dates, locations, and controlled status values.
- Identify missing values, invalid formats, and duplicate records.
- Document every approved transformation and exclusion.

## Data Protection
Avoid exposing direct identifiers in outputs. Preserve raw source data unchanged and write any approved derived data to the processed-data location.

## Review Criteria
Cleaning decisions should be reviewed for data quality, privacy impact, and potential effects on representativeness.
